```python
# Importing required libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Display settings
pd.set_option('display.max_columns', None)

print("All libraries imported successfully!")
```

```python
# Reading the Spotify dataset

df = pd.read_csv("dataset/dataset.csv")

df.head()
```

```python
# Display number of rows and columns

print("Dataset Shape:", df.shape)
print("Total Records:", df.shape[0])
print("Total Columns:", df.shape[1])
```

## 4. Data Understanding

In this section, we examine the structure and characteristics of the Spotify dataset. We explore column names, data types, non-null values and statistical summaries to understand the data before performing transformations.

```python
# Display first five records
display(df.head())

# Display last five records
display(df.tail())

# Display column names
print("Column Names:")
print(df.columns.tolist())

# Check dataset information
print("\nDataset Information:")
df.info()

# Generate statistical summary
print("\nStatistical Summary:")
display(df.describe())
```

## 5. Missing Value Analysis

Missing values can affect the accuracy of data analysis. In this section, we identify the number and percentage of missing values in each column.

```python
# Calculate missing values in each column
missing_values = df.isnull().sum()

# Calculate percentage of missing values
missing_percentage = (missing_values / len(df)) * 100

# Create a summary DataFrame
missing_summary = pd.DataFrame({
    'Missing Values': missing_values,
    'Missing Percentage': missing_percentage
})

display(missing_summary)
```

## 6. Data Cleaning and Transformation

Data cleaning and transformation are important steps in preparing a dataset for analysis. In this section, we remove duplicate records, eliminate unnecessary columns, convert data types where required and create new columns for further analysis.

```python
# Create a copy of the original dataset
df_clean = df.copy()

# Check duplicate records
print("Duplicate Records:", df_clean.duplicated().sum())

# Remove duplicate records
df_clean.drop_duplicates(inplace=True)

# Drop unnecessary identifier column
if 'track_id' in df_clean.columns:
    df_clean.drop(columns=['track_id'], inplace=True)

print("\nDataset Shape After Cleaning:", df_clean.shape)
```

```python
# Convert duration from milliseconds to minutes
if 'duration_ms' in df_clean.columns:
    df_clean['duration_minutes'] = df_clean['duration_ms'] / 60000

# Display updated dataset
display(df_clean.head())
```

```python
# Display data types after transformation
print(df_clean.dtypes)
```

## 7. Handling Missing Values Using Mean Imputation

Mean imputation is a data-cleaning technique used to replace missing numerical values with the mean of their respective columns.

In this section, we identify numerical columns containing missing values, calculate their mean and replace the missing values wherever applicable.

```python
# Identify numerical columns
numerical_columns = df_clean.select_dtypes(include=np.number).columns

print("Numerical Columns:")
print(numerical_columns.tolist())
```

```python
# Calculate missing values before imputation
print("Missing values before imputation:")
print(df_clean[numerical_columns].isnull().sum())

# Replace missing numerical values with column means
for col in numerical_columns:
    if df_clean[col].isnull().any():
        mean_value = df_clean[col].mean()
        df_clean[col] = df_clean[col].fillna(mean_value)

print("\nMissing values after mean imputation:")
print(df_clean[numerical_columns].isnull().sum())
```

```python
# Check remaining missing values in the dataset
print("Total Missing Values:", df_clean.isnull().sum().sum())
```

## 8. Exploratory Data Analysis (EDA)

Exploratory Data Analysis is the process of examining and analyzing a dataset to discover patterns, relationships and trends. In this section, we explore the popularity, energy, danceability and duration of Spotify tracks.

```python
# Calculate average values of numerical attributes

display(df_clean.describe().T)
```

```python
# Display the 10 most popular tracks

top_songs = df_clean.sort_values(
    by='popularity',
    ascending=False
).head(10)

display(top_songs[['track_name', 'artists', 'popularity']])
```

```python
# Calculate average popularity for each genre

genre_popularity = df_clean.groupby(
    'track_genre'
)['popularity'].mean().sort_values(ascending=False)

display(genre_popularity.head(10))
```

## 9. Data Visualization

Data visualization helps represent complex information through graphical formats. In this section, we use Matplotlib and Seaborn to visualize Spotify music attributes and identify patterns in the dataset.

```python
plt.figure(figsize=(10, 5))

sns.histplot(data=df_clean, x='popularity', bins=30, kde=True)

plt.title('Distribution of Spotify Track Popularity')
plt.xlabel('Popularity')
plt.ylabel('Number of Tracks')

plt.show()
```

```python
plt.figure(figsize=(12, 6))

sns.barplot(
    data=top_songs,
    x='popularity',
    y='track_name'
)

plt.title('Top 10 Most Popular Spotify Tracks')
plt.xlabel('Popularity')
plt.ylabel('Track Name')

plt.show()
```

```python
plt.figure(figsize=(10, 5))

sns.scatterplot(
    data=df_clean,
    x='danceability',
    y='popularity',
    alpha=0.4
)

plt.title('Danceability vs Track Popularity')
plt.xlabel('Danceability')
plt.ylabel('Popularity')

plt.show()
```

```python
plt.figure(figsize=(10, 5))

sns.histplot(data=df_clean, x='energy', bins=30, kde=True)

plt.title('Distribution of Spotify Track Energy')
plt.xlabel('Energy')
plt.ylabel('Number of Tracks')

plt.show()
```

```python
# Select numerical columns
numeric_df = df_clean.select_dtypes(include=np.number)

plt.figure(figsize=(12, 8))

sns.heatmap(
    numeric_df.corr(),
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Correlation Between Spotify Music Attributes')

plt.show()
```

## 10. Insight Generation

Based on the exploratory data analysis and visualizations, we identify meaningful patterns and relationships in Spotify music data.

```python
# Calculate important statistics

print("Average Track Popularity:",
      round(df_clean['popularity'].mean(), 2))

print("Average Danceability:",
      round(df_clean['danceability'].mean(), 2))

print("Average Energy:",
      round(df_clean['energy'].mean(), 2))

print("Average Track Duration (Minutes):",
      round(df_clean['duration_minutes'].mean(), 2))

print("\nTop 5 Genres by Average Popularity:")
display(genre_popularity.head(5))
```
